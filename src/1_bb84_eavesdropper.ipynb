```python
# Alice's random bits and random bases (0 = Z/standard, 1 = X/Hadamard)
alice_bits  = rng.integers(0, 2, n_bits)
alice_bases = rng.integers(0, 2, n_bits)
bob_bases   = rng.integers(0, 2, n_bits)
```

```python
def encode(bit, basis):
    """Prepare Alice's bit as a one-qubit circuit in the basis she chose.

    The X gate sets the bit. Basis 0 stays in Z; basis 1 adds H for the X basis.
    """
    qc = QuantumCircuit(1, 1)
    if bit == 1:
        qc.x(0)          # bit 1 flips |0> to |1>
    if basis == 1:
        qc.h(0)          # X basis: |0> becomes |+>, |1> becomes |->
    return qc
```

```python
def measure(qc, basis):
    """Measure a 1-qubit circuit in the given basis; return the classical bit."""
    if basis == 1:
        qc.h(0)                     # rotate X basis back to Z before measuring
    qc.measure(0, 0)
    counts = sim.run(qc, shots=1).result().get_counts()   # Aer runs x, h and measure as is: no transpile, much faster
    return int(max(counts, key=counts.get))

```

```python
def run_bb84(eavesdrop=False):
    """Send Alice's bits to Bob and return his measurements.

    With eavesdrop=True, Eve measures in a random basis and resends first.
    """
    bob_results = []
    for i in range(n_bits):
        qc = encode(alice_bits[i], alice_bases[i])
        if eavesdrop:
            # A wrong basis here is what Alice and Bob later see as errors.
            eve_basis = rng.integers(0, 2)
            eve_bit = measure(qc, eve_basis)
            qc = encode(eve_bit, eve_basis)
        bob_results.append(measure(qc, bob_bases[i]))
    return np.array(bob_results)
```

```python
def qber(bob_results):
    """Sift (keep matching bases) and return the error rate on the shared key.
    Real BB84 compares only a random sample of the sifted bits in public, then throws them away;
    here we compare every bit so the numbers are easy to see."""
    keep = alice_bases == bob_bases
    a = alice_bits[keep]
    b = bob_results[keep]
    if len(a) == 0:
        return 0.0
    return np.mean(a != b)

clean = run_bb84(eavesdrop=False)
print("QBER, no eavesdropper:", round(qber(clean), 3))

eve = run_bb84(eavesdrop=True)
print("QBER, with eavesdropper:", round(qber(eve), 3), "  (expect about 0.25; with 200 rounds, 0.18 to 0.33 is normal)")
```

### Hardware run: one job, not hundreds

The simulator cells above send one qubit per run, which is fine on a free local simulator. On real hardware that would be hundreds of queued, billed jobs. Instead, put 16 BB84 rounds side by side in one circuit (qubit *k* carries round *k*) and run it once with 200 shots.

Every shot repeats the same 16 rounds. That's fine here: Part 3 asks for the error rate caused by hardware noise, and 200 shots × the sifted rounds (6 of the 16 with this seed) give you about 1,200 measured bits to estimate it. They're 200 repeats of the same few states on the same few qubits, so treat the result as a rough estimate.

```python
def bb84_batch_circuit(bits, bases, bob_bases):
    """One circuit with one BB84 round per qubit. Reuses your encode() from above."""
    k = len(bits)
    qc = QuantumCircuit(k, k)
    for q in range(k):
        qc.compose(encode(bits[q], bases[q]), qubits=[q], clbits=[q], inplace=True)
        if bob_bases[q] == 1:
            qc.h(q)                 # Bob measures in the X basis
    qc.measure(range(k), range(k))
    return qc


def qber_from_counts(counts, bits, bases, bob_bases):
    """Error rate on the sifted rounds, over every shot."""
    keep = [q for q in range(len(bits)) if bases[q] == bob_bases[q]]
    errors = total = 0
    for bitstring, c in counts.items():
        bob = bitstring.zfill(len(bits))[::-1]   # Qiskit order: qubit 0 is the LAST character
        for q in keep:
            errors += c * (int(bob[q]) != bits[q])
            total += c
    return errors / total


k = 16                              # rounds per circuit = qubits used
hw_args = (alice_bits[:k], alice_bases[:k], bob_bases[:k])
hw_qc = bb84_batch_circuit(*hw_args)

# Check it on the local simulator first (free)
counts = sim.run(transpile(hw_qc, sim), shots=200).result().get_counts()
print("Batched QBER on the local simulator:", round(qber_from_counts(counts, *hw_args), 3))
```

## Authentication: QKD does not stop an imposter by itself

Part 4 of the problem asks what BB84 still needs from classical cryptography. We checked a simple case: an attacker who cuts the fiber and runs a clean key exchange with each side. Both error checks can look fine, and she can change a command in the middle. QKD’s error alarm does not fire. What stops her is a shared authentication tag (and a counter so she cannot replay an old real command).

```python
from pathlib import Path
import sys
_here = Path.cwd().resolve()
for _cand in (_here, _here.parent):
    if (_cand / "secure_grid").is_dir():
        if str(_cand) not in sys.path:
            sys.path.insert(0, str(_cand))
        break

from secure_grid.auth import Substation, tag

key = b"control-center/substation shared tag key"
msg = {"command": "CLOSE BKR 12", "counter": 41}
good = tag(key, msg)
fake = {"command": "OPEN  BKR 12", "counter": 41}
forged = tag(b"imposter's own key", fake)

sub = Substation(key, check_counter=True)
print("real tagged command:", sub.receive(msg, good))
print("forged tag:         ", sub.receive(fake, forged))
print("replay same counter:", sub.receive(msg, good))
```

## Part 4 — written answers

**Why secrecy comes from physics.** In BB84, a spy who measures a qubit has to guess the basis. Wrong guesses disturb the state, so Alice and Bob see extra errors when they compare a sample. Classical key exchange has no built-in “someone measured this” signal; its secrecy rests on math that a large quantum computer can break.

**What BB84 still needs.** The two ends still need a way to know they are talking to each other, not to an imposter who sits in the middle and runs two clean QKD sessions. That is authentication: a key or certificate installed ahead of time (and refreshed from the QKD key), or post-quantum signatures. Real QKD systems have also been attacked through imperfect detectors and side channels, so implementation matters.

**NSA / UK NCSC.** Both treat QKD as limited for national systems: it does not replace authentication, needs special fiber and trusted nodes over distance, and is not a substitute for migrating RSA/ECC to post-quantum cryptography on the wider network. We use QKD where tap *detection* is worth the cost, and post-quantum crypto on every link.
